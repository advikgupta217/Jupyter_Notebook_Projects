```python
subjects = ["Maths", "Science", "English", "Computer", "Hindi"]
marks = [85,93,93,84,95]

print("Subjects:", subjects)
print("Marks:", marks)

#Bar chart
plt.figure(figsize=(10,6))
plt.bar(subjects, marks)
plt.title("Student Marks by Subject")
plt.xlabel("Subjects")
plt.ylabel("Marks(out of 100)")
plt.ylim(0,100)
plt.grid(axis="y", linestyle="--", alpha=0.9)
plt.show()

#Line chart
plt.figure(figsize=(10,6))
plt.plot(subjects, marks, marker="^", linestyle="-")
plt.title("Student Marks Trend")
plt.xlabel("Subjects")
plt.ylabel("Marks(out of 100)")
plt.ylim(0,100)
plt.grid(True, linestyle="--", alpha=0.8)
plt.show()

#Pie chart
plt.figure(figsize=(7,7))
plt.pie(
    marks,
    labels=subjects,
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Contribution of Each Subject to Total Marks")
plt.show()

total=sum(marks)
average=total/len(marks)
highest=max(marks)
lowest=min(marks)

print("Total Marks:", total, "/ 500")
print("Average Marks:", average)
print("Highest Marks:", highest)
print("Lowest Marks:", lowest)
```

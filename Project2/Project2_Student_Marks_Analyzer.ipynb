```python
#Project 2 : Student Marks Analyzer

name = input("Enter student name: ")
subjects = ["Maths", "Science", "English", "Computer", "Hindi"]
marks = []

for subject in subjects:
    mark = float(input(f"Enter marks for {subject} (0-100): "))

    if mark<0 or mark>100:
        print("Invalid marks! Enter a value between 0 and 100.")
        break

    marks.append(mark)

if len(marks) == len(subjects):

    total = sum(marks)
    percentage = total / len(subjects)

    if percentage >= 90:
        grade = "A+"
    elif percentage >= 80:
        grade = "A"
    elif percentage >= 70:
        grade = "B"
    elif percentage >= 60:
        grade = "C"
    elif percentage >= 50:
        grade = "D"
    else:
        grade = "F"

    passed = all(mark >= 35 for mark in marks)

    print("\n======= STUDENT RESULT =======")
    print("Student Name:", name)

    for subject, mark in zip(subjects, marks):
        print(f"{subject}: {mark:g}")

    print("------------------------------")
    print(f"Total Marks: {total:g} / 500")
    print(f"Percentage: {percentage:.2f}%")
    print("Grade:", grade)
    print("Result:", "PASS" if passed else "FAIL")
    print("==============================")

else:
    print("Please restart the cell and enter valid marks.")
```
